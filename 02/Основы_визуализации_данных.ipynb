#### 2.1. Серии и датафреймы


Pandas — библиотека Python для обработки и анализа данных. Ее работа с данными строится поверх библиотеки NumPy, являющейся инструментом более низкого уровня. Предоставляет специальные структуры данных и операции для манипулирования числовыми таблицами и временными рядами.  Название происходит от английских слов «panel data» («панельные данные»), имеются в виду данные, полученные в результате исследований и структурированные в виде таблиц. С помощью Pandas специалисты в области аналитики данных могут группировать и визуализировать данные, создавать сводные таблицы и делать выборку по определенным признакам.

Основные структуры хранения данных в Pandas – Series и DataFrame.

*Series* (серии) – проиндексированный одномерный массив значений. Он похож на простой словарь, где имя элемента будет соответствовать индексу, а значение – значению записи. Индексом может быть числовой показатель (0, 1, 2 и т.д.), буквенные значения (a, b, с и т.д.) или другие данные. Если особое значение не задано, то числовые индексы проставляются автоматически.

*DataFrame* (датафрейм) — это проиндексированный многомерный массив значений: каждый столбец DataFrame является структурой Series. Похоже на электронную таблицу вроде Excel. В датафрейме можно также объединять данные в группы, сортировать по определенному признаку, производить вычисления.

Создать структуру Series и DataFrame можно на базе различных типов данных:

- словари Python;
- списки Python;
- массивы из numpy: ndarray;
- скалярные величины.

Но очень часто при работе с Pandas структуры не создаются, а используют готовые датафреймы.

#### 2.2. Подготовка данных для работы в Pandas


Что делать, если в датафрейме нашлись пустые ячейки? Это явно будет мешать вычислениям. Самое простое — оценить, насколько нужны столбцы или строки с незаполненными ячейками, и если нет в них необходимости — удалить.

Для упрощения работы можно удалить ненужные столбцы одним из следующих методов:

> **del df['A']** - удаляет столбец А из датафрема df

> **df.pop('A')** - удаляет строку А из датафрейма df

> **df.drop((['B', 'C'], axis=1))** - удаляет столбцы B и C из датафрейма df (для удаления строк нужно указывать axis=0).



Иногда можно перенести только нужные столбцы в копию датафрейма:  

> **new_df = df[['date', 'name']].copy()** - переносит в копию датафрейма столбцы date и name

Можно заполнить отсутствующие значения автоматически. Метод **dataframe.fillna ()** заполняет (заменяет) значения *Na* или *NaN* в dataframe с указанными значениями. NaN (Not a Value) - это недостающие данные.

Этот метод может быть использован для заполнения значений во всем датафрейме, в определенных столбцах или с учетом других условий.

Например,

> **df[['a', 'b']] = df[['a','b']].fillna(value=0)** - заменит все пропуски в столбцах a и b датафрейма df на нули.

> **df = df.fillna(df.mean())** заменит пропуски средними значениями соответствующего столбца;


Создадим для примера датафрейм с пропусками (они будут отображаться как **NaN**):

```python
data = 'stage,value,percent\n1,10,\n2,20,51\n3,30, 58 \n4, 40 \n,26'
```

```python
from io import StringIO
df = pd.read_csv(StringIO(data))
df
```

```python
df = df.fillna(df.mean()) # заполнение пропусков средними значениями
df
```

### 3.2. Matplotlib

Matplotlib — библиотека Python для визуализации данных двумерной графикой. Получаемые изображения могут быть использованы в качестве иллюстраций в публикациях. Matplotlib написан и поддерживался в основном Джоном Хантером. Matplotlib cостоит из множества модулей. Модули наполнены различными классами и функциями, которые иерархически связаны между собой.


Главным объектом при работе с matplotlib является рисунок (Figure), на котором располагаются одна или несколько областей рисования (Axes), элементы рисунка Artisits (заголовки, легенда и т.д.) и основа-холст (Canvas).

Интерфейс модуля pyplot позволяет выбирать готовые решения и настраивать базовые параметры рисунка.

Создать рисунок figure можно методом plt.figure(). После вызова любой графической команды, например, plt.scatter() или plt.plot(), всегда есть хотя бы одна область для рисования (по умолчанию прямоугольной формы).

Чтобы результат рисования отразился, используют plt.show().

Пример построения гистограммы:




```python
import matplotlib.pyplot as plt # загружаем модуль pyplot из matplotlib и назначаем ему псевдоним plt
fig = plt.figure() # создаем рисунок
plt.hist(d['Volume (ml)'], bins=30)  # формируем гистограмму из значений столбца 'Volume (ml)' датасета d c разбиением на 30 частей
plt.title('Объем (мл)') # подписываем рисунок
plt.grid(True) # просим показать фоновую сетку
```

Поменяем стиль изображения с помощью plt.style.use() - чтобы указать конкретный стиль, можно выбрать подходящий из галереи и скопировать его название: https://matplotlib.org/stable/gallery/style_sheets/style_sheets_reference.html

```python
plt.style.use('dark_background')
plt.hist(d['Volume (ml)'], bins=30)
plt.title('Объем (мл)')
plt.grid(True)
```

Список текстовых команд в pyplot:

plt.xlabel() – добавляет подпись оси OX;

plt.suptitle() – добавляет заголовок для рисунка Figure;

plt.annotate() – добавляет примечание, которое состоит из текста и необязательной стрелки в указанную область на рисунке.

### 3.3. Seaborn


**Seaborn** — популярная библиотека готовых шаблонов для статистической визуализации, написанная на основе matplotlib. В отличие от последнего в  Seaborn более выразительный высокоуровневый интерфейс, построение большинства простых графиков происходит в одну строчку кода, и сами графики выглядят тоже приятнее. Библиотека названа в честь Сэмюела Нормана Сиборна, героя сериала [The West Wing](https://ru.wikipedia.org/wiki/Западное_крыло_(телесериал)) (поэтому библиотеке принято назначать псевдоним sns - по инициалам персонажа - import seaborn as sns). Автор библиотеки — Michael Waskom, PhD, сотрудник Center for Neural Research (Нью-Йорк), выпускник Стэнфордского университета.


Воспользуемся датасетом  «Ирисы Фишера». Это классический учебный датасет, который встроен в seaborn.

Посмотрите, как выглядят эти ирисы на фото: https://ru.wikipedia.org/wiki/Ирисы_Фишера

Числовые столбцы показывают длину и ширину наружной и внутренней доли околоцветника для трех сортов ириса: setosa, virginica, versicolor.

```python
# Загружаем библиотеку seaborn и готовим данные:

import seaborn as sns
sns.set()
iris = sns.load_dataset("iris")
iris = iris[['petal_width', 'petal_length']] # оставим из датасета только два столбца - petal_width и petal_length

# Визуализация пар столбцов:
iris.plot.scatter(x='petal_width', y='petal_length')
```

Мы получили диаграмму рассеяния, которая указывает на сильную корреляцию между шириной и длиной лепестков (petal width и petal length ).

Для построения диаграмм рассеяния для каждой пары столбцов в датасете воспользуемся функцией pairplot из библиотеки seaborn.


```python

# Подготовка данных
import seaborn as sns
sns.set()
iris = sns.load_dataset("iris")
iris['species'] = iris['species'].replace({'setosa':0, 'versicolor':1, 'virginica':2}) #здесь категориальные признаки setosa, versicolor и virginica переводятся в числовые

# Визуализация всевозможных пар столбцов, по диагонали гистограммы распределения
sns.pairplot(iris)
```
